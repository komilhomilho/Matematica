**🔵EXERCÍCIO 1 - Custo de produção:**

Uma fábrica produz vigas de concreto. O custo total $C(x)$, em reais, para produzir $x$ vigas é dado por:

$C(x) = 5000 + 120x$

onde:

- R$ 5000,00  representam custos fixos e

- R$ 120,00 o custo por viga.

Pede-se:

a) Qual o custo para produzir 50 vigas?

b) Quantas vigas podem ser produzidas com um orçamento de R$ 11.000,00?

c) Forneça o correspondente código Python da solução dos itens a) e b) anteriores.

d) Faça o gráfico da função custo total  $C(x)$



**✅RESPOSTA**

<div align="justify">

✏️ **Resolução analítica**

**1.a)Qual o custo para produzir 50 vigas?**

1.a) Para $x = 50$:

$C(50) = 5000 + 120.50 = 5000 + 6000 = 11000$

**1.b) Quantas vigas podem ser produzidas com um orçamento de R$ 11.000?**

$5000 + 120x = 11000$

$120x = 6000 ⇒ x = 50$

**1.c) Forneça o correspondente código Python da solução dos itens a) e b) anteriores.**

🐍 **Código Python**

```python
def CustoTotal(x):
    return 5000 + 120 * x

def CalculoCustoVigia(x):
    return (x - 5000)/120


print(f"O Custo Total e de: R$ {CustoTotal(50):,.2f}")
print(f"A Quantidade de Vigias Que Podem Ser contratados é: {CalculoCustoVigia(11000):.0f}")
```

**1.d) Faça o gráfico da função custo total $C(x)$**

🐍 **Código Python**

```python
import numpy as np
import matplotlib.pyplot as plt
def CustoTotal(x):
    return 5000 + 120 * x

x = np.linspace(0, 500, 100)

y = CustoTotal(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label="Custo Total 5000 + 120x", color="green", linewidth=2.5)
plt.title("Gráfico de Custo Total")
plt.xlabel("Quantidade Produzida (x)")
plt.ylabel("Custo em R$ (y)")
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()

plt.show()
```

**🔵EXERCÍCIO 4 - Concentração de poluente**

A concentração de um poluente em um rio é modelada por:

$C(t) = -0,5t^{2} + 4t + 10$

onde *t* é o tempo em horas.

a) Quando a concentraçao é máxima?

b) Qual o valor máximo?

c) Faça um código em Python para fornecer as respostas dos itens a) e b) anteriores.

d) Faça o gráfico da função $C(t)$.




**✅RESPOSTA**

<div align="justify">

✏️ **Resolução analítica**

**4.a) Quando a concentraçao é máxima?**

$t_v = - \frac{b}{2a} = -\frac{4}{2(-0,5)} = 4$

**4.b) Qual o valor máximo?**

$C(4) = -0,5(16) + 16 + 10 = 18$


**4.c) Faça um código em Python para fornecer as respostas dos itens a) e b) anteriores.**

🐍 **Código Python**

```python
valores = {
    "a" : -0.5,
    "b" : 4,
    "c" : 10
}

def PoluicaoRio(t):
    return valores["a"] * t ** 2 + valores["b"] * t + valores["c"]

def ConcentracaoMaxima():
    return abs(valores["b"] / (2 * valores["a"]))

print(f"A Concentração Maxima de Poluente no rio é de: {ConcentracaoMaxima()}")
print(f"O valor Maximo Com O nivel de Concentração Maxima é de: {PoluicaoRio(ConcentracaoMaxima()):.2f}")
```

**d) Faça o gráfico da função $C(t)$.**

🐍 **Código Python**

```python


import numpy as np
import matplotlib.pyplot as plt

valores = {
    "a" : -0.5,
    "b" : 4,
    "c" : 10
}

def PoluicaoRio(t):
    return valores["a"] * t ** 2 + valores["b"] * t + valores["c"]

x = np.linspace(0, ConcentracaoMaxima() * 2, 100)

y = PoluicaoRio(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label="Nivel de Poluente No Rio", color="blue", linewidth=2.5)
plt.title("Concentração decorrente no tempo: ")
plt.xlabel("Concentração:")
plt.ylabel("Tempo (y):")
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()

plt.show()
```

**🔵EXERCÍCIO 5 - Lucro e produção**

O lucro de uma fábrica é dado por:

$L(x) = - 2x^{2}+80x- 500$

onde

$x$ é o número de unidades produzidas.

Pede-se:

a) Quantas unidades maximizam o lucro?

b) Qual o lucro máximo?

c) Faça um código Python para responder os itens a) e b) anteriores.

d) Faça o código Python para gerar o gráfico da função lucro $L(x)$.


**✅RESPOSTA**

<div align="justify">

✏️ **Resolução analítica**

**5.a) Quantas unidades maximizam o lucro?**

$x_v = - \frac{80}{2(-2)} = 20$

**5.b) Qual o lucro máximo?**

$L(20) = -2(400)+1600 - 500 = 300$

**5.c) Faça um código Python para responder os itens a) e b) anteriores.**

🐍 **Código Python**

```python
valores = {
    "a" : -2,
    "b" : 80,
    "c" : 500
}

def LucroFabrica(x):
    return valores['a'] * x ** 2 + valores["b"] * x - valores["c"]

def NumeroUnidadesLucroMaximo():
    return abs(valores["b"] / (2 * valores["a"]))

unidades_ideais = NumeroUnidadesLucroMaximo()
lucro_max = LucroFabrica(unidades_ideais)

print(f"O Valor De Unidades Necessária Para O Lucro Máximo é: {unidades_ideais:.0f}")
print(f"O Lucro Máximo da Fábrica é: R$ {lucro_max:,.2f}")
```

**5.d) Faça o código Pyhton para gerar o gráfico da função lucro $L(x)$.**

🐍 **Código Python**


```python
import numpy as np
import matplotlib.pyplot as plt

valores = {
    "a" : -2,
    "b" : 80,
    "c" : 500
}

def LucroFabrica(x):
    return valores['a'] * x ** 2 + valores["b"] * x - valores["c"]

def NumeroUnidadesLucroMaximo():
    return abs(valores["b"] / (2 * valores["a"]))

unidades_ideais = NumeroUnidadesLucroMaximo()
lucro_max = LucroFabrica(unidades_ideais)

x = np.linspace(0, NumeroUnidadesLucroMaximo() * 2, 100)
y = LucroFabrica(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label="Equação do Lucro (L(x))", color="blue", linewidth=2.5)
plt.plot(unidades_ideais, lucro_max, 'ro', label=f'Lucro Máximo (x={unidades_ideais:.0f})')
plt.title("Análise de Lucro da Fábrica")
plt.xlabel("Quantidade Produzida (x)")
plt.ylabel("Lucro em R$ (y)")

plt.axhline(0, color='black', linewidth=0.5, linestyle='--')
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()

plt.show()
```

**🔵EXERCÍCIO 6 - Desgaste de um componente mecânico ao longo do tempo**

O desgaste acumulado $D(t)$ de um componente mecânico sujeito a fadiga é modelado por uma função cúbica:

$D(t) = 0,2t^{3}-3t^{2}+12t$

onde:

$t$ é o tempo de operação em anos

$D(t)$ é o desgaste em milímetros

Pede-se:

a) Calcule o desgaste após 2 anos de operação.

b) Escreva o código em Python para o gráfico de $D(t)$.


**✅RESPOSTA**

<div align="justify">

✏️ **Resolução analítica**


**6.a) Calcule o desgaste após 2 anos de operação.**

Desgaste em $t = 2$

$D(t) = 0,2t^{3}-3t^{2}+12t$

$D(2)= 0,2(2)^{3} - 3(2)^{2}+12(2)$

$D(2)= 0,2(8) - 12+24$

$D(2)= 1,6 + 12$ mm

$D(2)= 13,6$ mm

**b) Escreva o código em Python para o gráfico de $D(t)$.**

🐍 **Código Python**


```python
import numpy as np
import matplotlib.pyplot as plt

valores = {
    "a" : 0.2,
    "b" : 3,
    "c" : 12
}

def Desgaste(t):
    return valores["a"] * t ** 3 - 3 * t ** 2 + 12 * t

print(f"O valor do Desgaste após 2 anos de Operação é: {Desgaste(2):.1f}mm")

x = np.linspace(0, 10, 100)

y = Desgaste(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label="Curva de Desgaste", color="blue", linewidth=1)
plt.title("Desgaste da Máquina ao Longo do Tempo")
plt.xlabel("Tempo em Anos (t)")
plt.ylabel("Desgaste em mm (y)")
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()

plt.show()
```

**7.b) Faça um código Python para exibir o gráfico desta função.**

🐍 **Código Python**

```python
import numpy as np
import matplotlib.pyplot as plt

def FormatarMonetario(valor):
    return f"R$ {valor:,.2f}"

def Custo(nivelProducao):
    if nivelProducao < 3:
        resultado = (10 * nivelProducao + 5) * 1000000
    elif nivelProducao == 3:
        resultado = 35 * 1000000
    else:
        resultado = (2 * nivelProducao ** 2 - 4 * nivelProducao + 11) * 1000000
    return resultado

x = np.linspace(0, 5, 41)

y = np.asarray([Custo(valor) for valor in x]) / 1000000

plt.figure(figsize=(8, 5))
plt.plot(x, y, label="Custo", color="red", linewidth=2.5)
plt.title("Custo em Milhões Com o Nivel de Produção")
plt.xlabel("Nivel De Produção:")
plt.ylabel("Custo (Milhões de R$)")
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()

plt.show()
```
