**Função não nativa**

Algumas funções **não são nativas do python** e precisam da importação de alguma bibioteca específica para executarem. Por exemplo, a função $g(x) = log_{10}(x)$, correspondendo ao logaritmo de base $10$.

Pode-se importá-la da biblioteca `numpy`. Em particular, para esta função $D(g) = \mathbb{R}_{+}^{*}$ e $Im(g) = \mathbb{R}$.

**📌EXEMPLO 2**

```python
# Imortação da biblioteca contendo funções matemáticas não nativas
import numpy as np

# Definindo uma função
def g(x):
    return np.log10(x)
```

```python
# Avaliando g(x) com alguns valores
print(g(10))
print(g(100))
print(g(1000))
print(g(84.5))
```

```python
# Avaliação fora do domínio da função g(x)
print(g(-1))
```

O erro obtido ao executar a célula anterior se deve ao fato de o domínio de $g(x) = log_{10}(x)$ **não comportar** valores negativos.

**Avaliando a função com um conjunto de valores**

Suponha que se queira avaliar a função $f(x) = x^{2}$ no conjunto $A = \{-1,0,1,2,3,5,10\}$. Neste caso usa-se o loop for:

**📌EXEMPLO 3**

```python
# Definindo uma função
def f(x):
    return x**2

# Definindo o conjunto de entrada A
A = {-1,0,1,2,3,5,10}

# Percorrendo o cojunto A
for x in A:
    print(f(x))
```

<div align="justify">

O Python desordena automaticamente os elementos — um conjunto (*set*) é uma coleção não ordenada. Por isso, o `for` percorre os elementos **em ordem arbitrária**, e não necessariamente na ordem que você escreveu. Para garantir que os valores serão impressos na ordem crescente, usa-se `sorted` que os ordena:
</br>

```python
# Percorrendo o cojunto A em ordem crescente, use o comando sorted()
for x in sorted(A):
    print(f(x))
```

Para um espaçamento não inteiro pode-se usar `arange` da biblioteca numpy. Por exemplo, considere o conjunto $A = \{0,0.1,0.2, ..., 2\}$ e veja o código a seguir:

**📌EXEMPLO 4**

```python
# Imortação da biblioteca contendo funções matemáticas não nativas
import numpy as np

# Definindo uma função
def f(x):
    return x**2

# Gerando o conjunto de entrada A
A = np.arange(0,2.1,0.1) # O último elemento é desconsiderado.Ele percorrerá os valores de 0 até 2.0 e o valor 2.1 será ignorado

# Percorrendo o cojunto A
for x in A:
    print(f(x))
```

**Operações com Funções**

<div align="justify">

Com duas funções pode-se efetuar operações como soma, mutiplicação, e compor seus resultados (saída de uma funcionando como entrada de outra).

image.png

**📌EXEMPLO 5 - Considere as seguintes funções:**

$f(x) = 2x$


$g(x) = 3x + x^{2}$


$h(x) =2^{x}$

```python
# Definindo as funções
def f(x):
    return 2*x

def g(x):
    return 3*x + x**2

def h(x):
    return 2**x
```

```python
# Imprimindo alguns valores
print(f(2))
print(g(-1))
print(h(3))
```

**✔️Somar duas funções:** $f(x)+g(x)$

$f(x)+g(x) = 2x + 3x + x^2$

$f(x)+g(x) = 5x + x^2$

Se $x = 2$, esta soma ficará:

$f(2)+g(2) = 5(2) + (2)^2$

$f(2)+g(2) = 14$

```python
# Função soma de f com g
def f_mais_g(x):
    return f(x) + g(x)
# Somando f(x)+ g(x) quando x = 2.
print(f_mais_g(2))
```

---

⚠️⚠️⚠️⚠️ **ATENÇÃO!!!!!!!** ⚠️⚠️⚠️⚠️

O alinhamento (Identação) em Python é
**IMPORTANTE**!!!!!!!!!
Se o comando `print` acima for colocado alinhado com o `return`, **NÃO IRÁ IMPRIMIR** o resultado 14 pois o Python entenderá que ele é parte da função `f_mais_g(x)`.

---

**✔️Multiplicar duas funções:** $g(x)*h(x)$

$g(x)*h(x) = (3x + x^{2})*2^{x}$

Se $x = -3$, esta multiplicação ficará:

$g(-3)*h(-3) = (3(-3) + (-3)^{2})*2^{-3}$

$g(-3)*h(-3) = ((-9) + 9)*2^{-3}$

$g(-3)*h(-3) = (0)*2^{-3}$

$g(-3)*h(-3) = 0$


```python
# Função produto de g e h
def g_produto_h(x):
    return g(x)*h(x)

# Multiplicando g(x)*h(x) quando x = -3.
print(g_produto_h(-3))
```

 **✔️Função Composta** $fog$ **ou**  $f(g(x))$

$fog(x) = f(g(x))$

$fog(x) = 2(3x + x^{2})$

$fog(x) = 6x + 2x^{2}$

Se $x = 1$, esta função ficará:

$fog(1) = 6(1) + 2(1)^{2}$

$fog(1) = 6 + 2$

$fog(1) = 8$

```python
# Função composta
def f_composta_g(x):
    return f(g(x))   # A saída da g funciona como a entrada de f
print(f_composta_g(1))
```

**Gráfico de uma Função**

<div align="justify">

O gráfico de uma função é uma importante ferramenta visual para estudá-la. Em muitas situações analisamos gráficos de funções para obtermos informações sobre elas e suas implicações.

Por exemplo, o gráfico da função $f(x) = 2x$, plotado de 0 até 3, com espaçamento de 0.1, é apresentado na célula a seguir:

```python
# Importação de bibliotecas
import numpy as np
import matplotlib.pyplot as plt # Biblioteca necessária para o esboço de gráficos

# Definindo uma função
def f(x):
    return 2*x

# Gerando o conjunto de entrada
x = np.linspace(0, 3, 31)  # uma outra forma de plotar

# Plotando o gráfico
plt.plot(x,f(x),'.')
plt.show()
```

---
🚨🚨🚨🚨🚨🚨 **IMPORTANTE** 🚨🚨🚨🚨🚨🚨

o NumPy gera exatamente:

0.0, 0.1, 0.2, 0.3, ..., 2.9, 3.0

Ou seja, **31 valores espaçados por 0.1** e representados pelos pontos em azul.

Mas o Matplotlib, por padrão, decide **automaticamente** onde colocar as marcações do eixo (*ticks*).
Ele escolhe intervalos visualmente “bons”: 0.5, 1.0, 1.5 etc.

---

Pode-se incrementar o gráfico anterior colocando outros elementos como título, legenda, etc.:

```python
# Importação de bibliotecas
import numpy as np
import matplotlib.pyplot as plt # Biblioteca necessária para o esboço de gráficos

# Definindo uma função
def f(x):
    return 2*x

# Gerando o conjunto de entrada
x = np.linspace(0, 3, 31)

# Configurações adicionais
plt.title('Título')
plt.xlabel('Variável 1')
plt.ylabel('Variável 2')
plt.grid() # Grade


# Plotando o gráfico
plt.plot(x,f(x),'-',color='red') # '-' muda para uma linha ligando os pontos de cor vermelha
plt.show()
```

Portanto, no código anterior, em Configurações adicionais, pode-se alterar para:

```python
# Importação de bibliotecas
import numpy as np
import matplotlib.pyplot as plt # Biblioteca necessária para o esboço de gráficos

# Definindo uma função
def f(x):
    return 2*x

# Gerando o conjunto de entrada
x = np.linspace(0, 3, 31)

# Configurações adicionais
plt.title('Gráfico da Função f(x) = 2x')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid() # Grade

# Plotando o gráfico
plt.plot(x,f(x),'-',color='blue') # voltando a cor azul
plt.show()
```

**Outras funções e seus gráficos**

**📌EXEMPLO 6**

Esboce o gráfico da seguinte função:

image.png

```python
import matplotlib.pyplot as plt

# Criando os valores de x
x = list(range(-10, 11))  # de -10 até 11, outra forma de definir os valores de x

# Definindo f(x) = |x|
y = [abs(i) for i in x] # a funçao abs(x), calcula o módulo de x. Neste comando i percorre todos os valores definidos para x anteriormente.

# Criando o gráfico
plt.plot(x, y, label='f(x) = |x|', color='blue')

# Adicionando título e rótulos
plt.title('Gráfico da função f(x) = |x|')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()

# Mostrando o gráfico
plt.show()
```

**Função $sen x$**

```python
import matplotlib.pyplot as plt
import numpy as np

# Criando valores de x de -2π a 2π
x = np.linspace(-2*np.pi, 2*np.pi, 200)

# Calculando f(x) = sen(x)
y = np.sin(x)

# Criando o gráfico
plt.plot(x, y, label='f(x) = sen(x)', color='blue')

# Criando os ticks do eixo X (múltiplos de pi)
xticks = [
    -2*np.pi, -3*np.pi/2, -np.pi, -np.pi/2,
    0,
    np.pi/2, np.pi, 3*np.pi/2, 2*np.pi
]

xtick_labels = [
    r'$-2\pi$', r'$-\frac{3\pi}{2}$', r'$-\pi$', r'$-\frac{\pi}{2}$',
    r'$0$',
    r'$\frac{\pi}{2}$', r'$\pi$', r'$\frac{3\pi}{2}$', r'$2\pi$'
]

plt.xticks(xticks, xtick_labels)

# Adicionando título e rótulos
plt.title('Gráfico da função f(x) = sen(x)')
plt.xlabel('x (radianos)')
plt.ylabel('sen(x)')
plt.grid(True)
plt.legend()

# Mostrando o gráfico
plt.show()
```

**Função $tg x$**

```python
import matplotlib.pyplot as plt
import numpy as np

# Criando valores de x de -2π a 2π
x = np.linspace(-2*np.pi, 2*np.pi, 400)
# Usando a função tg(x) da biblioteca numpy
y = np.tan(x)

# Limitando o eixo y para evitar valores muito grandes
plt.ylim(-10, 10)

# Criando o gráfico
plt.plot(x, y, label='f(x) = tan(x)', color='green')

# Criando os ticks do eixo X (múltiplos de pi)
xticks = [
    -2*np.pi, -3*np.pi/2, -np.pi, -np.pi/2,
    0,
    np.pi/2, np.pi, 3*np.pi/2, 2*np.pi
]

xtick_labels = [
    r'$-2\pi$', r'$-\frac{3\pi}{2}$', r'$-\pi$', r'$-\frac{\pi}{2}$',
    r'$0$',
    r'$\frac{\pi}{2}$', r'$\pi$', r'$\frac{3\pi}{2}$', r'$2\pi$'
]

plt.xticks(xticks, xtick_labels)

# Título e rótulos
plt.title('Gráfico da função f(x) = tg(x)')
plt.xlabel('x (radianos)')
plt.ylabel('tg(x)')
plt.grid(True)
plt.legend()

# Mostrar o gráfico
plt.show()
```

**Uma Animaçao com o Gráfico da Tangente**

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

plt.rcParams['animation.embed_limit'] = 50  # aumenta limite

# Intervalo de x
x = np.linspace(-2*np.pi, 2*np.pi, 300)   # reduzido para caber no limite
y = np.tan(x)

fig, ax = plt.subplots(figsize=(8, 5))

ax.set_xlim(-2*np.pi, 2*np.pi)
ax.set_ylim(-10, 10)
ax.grid(True)
ax.set_title("Animação da função f(x) = tg(x)")
ax.set_xlabel("x (radianos)")
ax.set_ylabel("tg(x)")

linha, = ax.plot([], [], color='green', lw=2)

def init():
    linha.set_data([], [])
    return linha,

def update(frame):
    linha.set_data(x[:frame], y[:frame])
    return linha,

anim = FuncAnimation(
    fig, update, frames=len(x),
    init_func=init, interval=30, blit=True # interval = 10, isto é abaixo de 30, aumentará a velocidade
)

HTML(anim.to_jshtml())
```
